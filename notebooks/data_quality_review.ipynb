# PostgreSQL data quality review
Read-only SQL companion to the framework and register. Requires the existing local PostgreSQL warehouse and project Python environment. Credentials remain in ignored .env.

```python
from pathlib import Path
import sys
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(root / 'src'))
from warehouse.db import connect
```

Rule summary

```python
query = 'SELECT * FROM quality.v_latest_dq_summary ORDER BY quality_dimension'
with connect() as conn:
    for row in conn.execute(query):
        print(row)
```

Mapping coverage

```python
query = 'SELECT * FROM quality.v_latest_coverage ORDER BY source_name,source_month'
with connect() as conn:
    for row in conn.execute(query):
        print(row)
```

Run history

```python
query = 'SELECT run_id,as_of_date,status,started_at,completed_at FROM quality.dq_run ORDER BY run_id'
with connect() as conn:
    for row in conn.execute(query):
        print(row)
```
